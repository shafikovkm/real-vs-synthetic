Векторные графики сохранены в `figures/categories.pdf` и `figures/lengths.pdf`. Укажите рассчитанную долю контекста и статистику длин в окончательной версии отчёта.

```python
import matplotlib.pyplot as plt
from collections import Counter
rows = [json.loads(line) for line in Path("data/databricks-dolly-15k.jsonl").read_text(encoding="utf-8").splitlines()]
counts = Counter(row["category"] for row in rows)
fig, ax = plt.subplots(figsize=(9, 4))
ax.barh(list(sorted(counts)), [counts[k] for k in sorted(counts)])
ax.set_title("Dolly: category distribution"); ax.set_xlabel("Examples"); plt.show()
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
for ax, field in zip(axes, ("instruction", "response")):
    ax.hist([len(row[field].split()) for row in rows], bins=50)
    ax.set_xscale("symlog", linthresh=1); ax.set_title(field); ax.set_xlabel("Words (log scale)")
plt.tight_layout(); plt.show()
```
